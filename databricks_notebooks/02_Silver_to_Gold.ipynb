# 🥇 Stratul Gold: Analiza Pieței Imobiliare (Stil dbt)
**Obiectiv:** Agregarea datelor curate din stratul Silver pentru a crea rapoarte de business finale, exclusiv în limbaj SQL.

```python
%sql

-- 💾 SALVARE STRAT GOLD: Creăm un tabel permanent în baza de date default numit 'gold_top_cartiere'
CREATE OR REPLACE TABLE default.gold_top_cartiere AS

WITH raw_silver AS (
    SELECT * FROM delta.`/Volumes/workspace/default/raw_data/silver_storia/`
),


calculated_metrics AS (
    SELECT 
        COALESCE(City_Sector, 'Ilfov / Necunoscut') AS Sector,
        Neighborhood AS Cartier,
        
        -- Calculăm mediile ignorând automat valorile null
        COUNT(listing_id) AS Numar_Anunturi,
        ROUND(AVG(Price_EUR), 0) AS Pret_Mediu_Total_EUR,
        ROUND(AVG(Price_per_sqm), 0) AS Pret_Mediu_MP_EUR,
        ROUND(AVG(rooms), 1) AS Numar_Mediu_Camere

    FROM raw_silver
    -- Excludem anomaliile complet goale
    WHERE Neighborhood IS NOT NULL AND Neighborhood != ''
    
    GROUP BY 
        City_Sector, 
        Neighborhood
)


SELECT * FROM calculated_metrics
WHERE Numar_Anunturi >= 2
ORDER BY Pret_Mediu_MP_EUR DESC
```
